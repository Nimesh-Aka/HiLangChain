# Model Integration and Streaming and Batching

```python
import os
from dotenv import load_dotenv

load_dotenv()

os.environ["GROQ_API_KEY"] = os.getenv("GROQ_API_KEY") 
```

```python
from langchain_groq import ChatGroq

model = ChatGroq(
    model="openai/gpt-oss-20b"
)
response = model.invoke("Why NBA is so famous?")
response
```

### Streaming & Batching 

Streaming means receiving the LLM's response piece by piece as it is being generated, instead of waiting for the entire response.

```python
for chunk in model.stream("Why NBA is so famous?"):
    print(chunk.text, end="!", flush=True)
```

### Batching

Batching means processing multiple inputs together in a single batch, rather than processing each input individually.

```python
model.batch(
    [
        "Why NBA is so famous?",
        "Who is the president of USA?",
        "What is the weather in Matara?"
    ],
    config={
        'max_concurrency': 5,
    }
)
```
