
# 🧠 Understanding Prompts and Messages in LangChain

## 1. Types of Prompts

| Concept | Meaning | Example |
|---|---|---|
| Text Prompt | Plain text sent to an LLM | `"Explain Python"` |
| Message Prompt | Structured messages with roles | System + Human messages |

## 2. Types of Messages

| Message | Responsibility |
|---|---|
| System Message | Defines AI behaviour and instructions |
| Human Message | Contains the user's input |
| AI Message | Contains the AI's response or tool-call request |
| Tool Message | Contains the result returned by a tool |

## 3. How Messages Work Together

```mermaid
flowchart TD
    A["User asks a question"] --> B["Human Message"]
    B --> C["LLM receives messages"]
    D["System Message"] --> C
    C --> E["AI Message"]
    E --> F{"Does AI need a tool?"}
    F -->|No| G["Final AI Response"]
    F -->|Yes| H["AI Tool Call"]
    H --> I["Tool Executes"]
    I --> J["Tool Message"]
    J --> K["LLM receives tool result"]
    K --> G
```

## 4. Example Conversation

**System Message:**
You are a helpful AI assistant.

**Human Message:**
What is the weather in Matara?

**AI Message (Tool Call):**
Call `get_weather(location="Matara")`.

**Tool Message:**
Sunny with a temperature of 25°C.

**AI Message (Final Response):**
The weather in Matara is sunny, with a temperature of 25°C.

## 5. Quick Summary

- Text Prompt → Plain text input.
- Message Prompt → Structured conversation input.
- System → AI instructions.
- Human → User input.
- AI → AI response or tool request.
- Tool → Tool execution result.
